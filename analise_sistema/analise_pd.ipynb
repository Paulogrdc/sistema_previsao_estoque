```python
# inspeção 

display(df_movimentacao)
display(df_movimentacao.head(15))
display(list(df_movimentacao.columns))
display(df_movimentacao.info())
display(df_movimentacao.describe())
```

```python
# Tratamento

# tratatar a tabeala data
df_movimentacao["data"] = pd.to_datetime(df_movimentacao["data"], format="%Y-%m-%d %T")

# ver se tem valores duplicados 
df_movimentacao = df_movimentacao.drop_duplicates(subset=["id_importacao"])


display(df_movimentacao.info())
display(df_movimentacao)
```

```python
# Criar colunas  

df_analise = df_movimentacao

df_analise["Mes-ano"] = df_movimentacao["data"].dt.to_period("M")


# 

display(df_analise)
```
